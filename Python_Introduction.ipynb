```python
# The partition() method searches for a specified string & splits the string into tuple containing three elements.

# The first element contains the part before the arguement string.
# The second element contains the arguement string.
# The third element contains the part after arguement string.

str5 = "Natural language processing with Python and R and Java"
L = str5.partition("and")
print(L)
```

```python
# The rpartition() method searches for the last occurance of the specified string & splits the string into the tuple
# containing three elements.

# The first element contains the part before the arguement string.
# The second element contains the arguement string.
# The third element contains the part after arguement string.

str6 = "Natural language processing with Python and R and Java"
L = str6.rpartition("and")
print(L)
```
